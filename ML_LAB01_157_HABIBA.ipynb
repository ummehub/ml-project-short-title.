```python
import numpy as np

np.random.seed(42)
marks = np.array([55, 72, 48, 90, 66, 81])

print('Marks:', marks)
print('Mean:', marks.mean())
print('Highest:', marks.max())
print('Passed:', marks[marks >= 50])

print('Minimum:', marks.min())
print('Standard deviation:', marks.std())

bonus_marks = marks + 5
print('Bonus marks:', bonus_marks)
print('Original marks:', marks)

print('Number passed:', np.sum(marks >= 50))

matrix = marks.reshape(2, 3)
print('2 × 3 matrix:')
print(matrix)
```

```python
import numpy as np

marks = np.array([45, 67, 32, 80, 55, 39])

minimum = np.min(marks)
std_dev = np.std(marks)
print("Minimum mark:", minimum)
print("Standard deviation:", std_dev)

bonus_marks = marks + 5
print("Original marks:", marks)
print("Marks after bonus:", bonus_marks)

passed = np.sum(marks >= 40)
print("Number of students passed:", passed)

matrix = marks.reshape(2, 3)
print("2 × 3 Matrix:")
print(matrix)
```

```python
import random
import numpy as np

def generate_scores(use_seed):
    if use_seed:
        random.seed(42)
        np.random.seed(42)
    return np.random.randint(40, 101, size=5)

print('Seeded run 1:', generate_scores(True))
print('Seeded run 2:', generate_scores(True))
print('Unseeded run 1:', generate_scores(False))
print('Unseeded run 2:', generate_scores(False))
```
